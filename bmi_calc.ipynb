```python
from langgraph.graph import StateGraph
from langgraph.constants import START, END
from typing import TypedDict
```

```python
class BMI_State(TypedDict):
    weight: float
    height: float
    bmi: float
```

```python
def calculate_bmi(state: BMI_State) -> BMI_State:
    weight = state["weight"]
    height = state["height"]
    bmi = weight / (height ** 2)
    state["bmi"] = bmi
    return state
```

```python
graph = StateGraph(BMI_State)

# Add nodes to the graph
graph.add_node("calculate_bmi", calculate_bmi)

# Add edges to the graph
graph.add_edge(START, "calculate_bmi")
graph.add_edge("calculate_bmi", END)

# Compile the graph
workflow = graph.compile()

# Execute the graph with an initial state
initial_state = {"weight": 70, "height": 1.75, "bmi": 0}
final_state = workflow.invoke(initial_state)

print(f"BMI: {final_state['bmi']}")  # Output: 22.857142857142858
```
